```python
# Install the modern official Kaggle library if not already present
!pip install -q kagglehub

import kagglehub
import os

print("Status: Initializing automated handshake with Kaggle servers...")

# Download the latest production-grade release of the PlantVillage Potato dataset
path = kagglehub.dataset_download("aarishasifkhan/plantvillage-potato-disease-dataset")

print("\n=== SYSTEM LOG ===")
print("Connection Status: Success! Dataset pulled straight to local cloud workspace.")
print("Path to dataset files stored in 'path' variable:", path)

# Read and print the folders inside the path to verify the contents
print("Sub-directories discovered inside the dataset source:", os.listdir(path))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import os

# Set standard pixel size dimensions and model batch constraints
IMG_SIZE = 256
BATCH_SIZE = 32

# Map the active source directory straight inside the true 'PlantVillage' folder
ACTIVE_DATA_DIR = os.path.join(path, 'PlantVillage')

print(f"Status: Accessing image assets from mapped source: {ACTIVE_DATA_DIR}")

# Create the 80% Training Split Data Engine
train_ds = tf.keras.utils.image_dataset_from_directory(
    ACTIVE_DATA_DIR,                   # Pointing directly to the deep sub-directory
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)

# Create the 20% Validation/Testing Split Data Engine
val_ds = tf.keras.utils.image_dataset_from_directory(
    ACTIVE_DATA_DIR,                   # Pointing directly to the deep sub-directory
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)

# Extract and lock the actual classification diagnostic target text array names
class_names = train_ds.class_names

print("\n=== SYSTEM LOG ===")
print("Data Pipeline Status: Operational & Successfully Hooked!")
print("Verified Crop Disease Classes Discovered:", class_names)
```

```python
from tensorflow.keras import layers, models

print("Status: Building the mathematical layer stack for the Neural Network...")

# Construct the Sequential Convolutional Neural Network
model = models.Sequential([
    # Layer 1: Normalization - standardizes raw pixel values from 0-255 down to 0.0 - 1.0
    layers.Rescaling(1./255, input_shape=(IMG_SIZE, IMG_SIZE, 3)),

    # Layer 2 & 3: First Convolutional Block - Extracts simple features like leaf edges and veins
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Layer 4 & 5: Second Convolutional Block - Extracts complex structures like dark spots or mold patterns
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    # Layer 6: Flattening - Converts the 2D feature map matrix into a long 1D numeric vector
    layers.Flatten(),

    # Layer 7: Dense Hidden Layer - Connects extracted patterns to perform logical decision mapping
    layers.Dense(64, activation='relu'),

    # Layer 8: Dense Output Layer - Uses Softmax activation to output probabilities for our 3 target classes
    layers.Dense(len(class_names), activation='softmax')
])

# Compile the neural architecture using the modern 'Adam' Gradient Descent optimizer
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\n=== SYSTEM LOG ===")
print("Model Status: Architecture fully defined, compiled, and locked!")
print("Printing Model Structural Summary Table:")
print("-" * 50)
model.summary()
```

```python
print("Status: Initializing weights optimization via GPU acceleration...")
print("Please wait while the neural network scans the image structures...\n")

# Execute the model fitting process over 3 complete training epochs
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=3
)

print("\n=== SYSTEM LOG ===")
print("Project 1 Engine Status: Training finalized and successfully saved!")
```

```python
print("Status: Evaluating Image Classification Model Performance...")

# Evaluate the trained image classification model on the validation dataset
loss, accuracy = model.evaluate(val_ds)

print("\n=== MODEL PERFORMANCE METRICS ===")
print(f"1. Validation Loss: {loss:.4f}")
print(f"2. Validation Accuracy: {accuracy:.4f} (Or {accuracy * 100:.2f}%)")
```

```python
import os

print("Status: Commencing model serialization...")

# Create a local path name for your model architecture
MODEL_NAME = "crop_disease_cnn_model.keras"

# Save the complete trained network structural weights
model.save(MODEL_NAME)

print("\n=== SYSTEM LOG ===")
print(f"Success! Model weights exported and saved locally as: {MODEL_NAME}")
print("Checking file size on disk:", os.path.getsize(MODEL_NAME) / (1024 * 1024), "MB")

```
