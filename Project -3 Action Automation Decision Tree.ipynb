```python
import numpy as np
import pandas as pd

print("Status: Launching Synthetic Automation Data Generation via NumPy...")

# 1. Generate 400 historical agricultural recommendation records using NumPy
np.random.seed(42)
num_records = 400

# Feature 1: Simulating Predicted Disease IDs from Project 1
# (0 = Early Blight, 1 = Late Blight, 2 = Healthy)
# We set probabilities (p) so 40% is early blight, 40% is late blight, and 20% is healthy
disease_id = np.random.choice([0, 1, 2], size=num_records, p=[0.4, 0.4, 0.2])

# Feature 2: Simulating Predicted Damage Severity Percentages from Project 2
# Generates continuous numeric loss percentages ranging from 0.0% to 100.0%
severity_scores = np.random.uniform(0.0, 100.0, num_records)

# Target Y: Simulating the True Recommended Action Code that a botanist would give
# Code 0 = Apply Industrial Copper Fungicide (High Risk / Urgent)
# Code 1 = Apply Organic Neem Spray (Low Risk / Warning Alert)
# Code 2 = No Chemical Action Required / Regular Monitoring (System Normal)
action_labels = np.zeros(num_records, dtype=np.int32)

# Building a logical rule mapping loop to align the synthetic relationships
for i in range(num_records):
    if disease_id[i] == 2:  # If plant is healthy
        action_labels[i] = 2 # Action: No action required
    elif disease_id[i] == 1 and severity_scores[i] > 40.0:  # Severe Late Blight case
        action_labels[i] = 0 # Action: Urgent chemical spray
    elif disease_id[i] == 0 and severity_scores[i] > 50.0:  # Severe Early Blight case
        action_labels[i] = 0 # Action: Urgent chemical spray
    else:
        action_labels[i] = 1 # Low-severity warning case

# 2. Binding separate NumPy arrays into a highly structured Pandas DataFrame
automation_dataframe = pd.DataFrame({
    'Detected_Disease_ID': disease_id,
    'Severity_Loss_Percent': severity_scores,
    'Prescription_Action_Code': action_labels
})

print("\n=== SUCCESS: PRODUCTION AUTOMATION DATAFRAME GENERATED ===")
print("Displaying the top 5 matrix rows of your pipeline controller dataset:")
print("-" * 75)
print(automation_dataframe.head()) # Renders the structured data grid screen output
```

```python
from sklearn.tree import DecisionTreeClassifier
import numpy as np

print("Status: Splitting data and initializing Supervised Decision Tree...")

# 1. Isolate the operational Input Features (X) and discrete Target Labels (y)
X_tree = automation_dataframe[['Detected_Disease_ID', 'Severity_Loss_Percent']]
y_tree = automation_dataframe['Prescription_Action_Code']

# 2. Build the Decision Tree model using Entropy math constraints to find structural splits
# We cap the depth at 4 layers to keep the logic clean and prevent overfitting
automator_model = DecisionTreeClassifier(criterion='entropy', max_depth=4, random_state=42)

# 3. Train the model state using the standard fit workflow
automator_model.fit(X_tree, y_tree)

# 4. Evaluate classification accuracy performance score
tree_accuracy = automator_model.score(X_tree, y_tree)

print("\n=== SYSTEM LOG ===")
print("Model Status: Decision Tree Classifier training finalized successfully!")
print(f"Calculated Training Accuracy Score: {tree_accuracy * 100:.2f}%\n")

print("-" * 75)
print("EXECUTING LIVE SYSTEM CONTROL PIPELINE TEST SCENARIO:")

# Live Simulated Test Input Matrix:
# Let's assume Project 1 CNN diagnosed Late Blight (ID = 1)
# and Project 2 Regression calculated a high yield destruction score of 65.50%
live_unseen_inputs = np.array([[1, 65.5]])

# Use our trained Decision Tree brain to predict the action classification code
predicted_action_code = automator_model.predict(live_unseen_inputs)[0]

# Mapping the numeric output class indices back to actual clear botanical actions
actions_dictionary = {
    0: "CRITICAL ALERT (CODE 0): Apply Industrial Copper Fungicide within 24 hours.",
    1: "WARNING UPDATE (CODE 1): Apply Organic Neem Spray and decrease irrigation frequency.",
    2: "SYSTEM NORMAL  (CODE 2): Crop matrix is verified healthy. Continuous automated monitoring active."
}

print(f"1. Input Vector Fed to Pipeline Engine: [Disease_ID=1, Severity=65.5%]")
print(f"2. Automated Prescription Output: {actions_dictionary[predicted_action_code]}")
print("-" * 75)
```
