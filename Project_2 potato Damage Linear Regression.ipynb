```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

print("Status: Initializing Tabular Data Generation via NumPy...")

# 1. Generate an industrial-scale agricultural dataset using NumPy arrays (500 data rows)
np.random.seed(42)
num_farms = 500

# Generating raw environmental numbers (Features X) using NumPy
temperature = np.random.uniform(15.0, 42.0, num_farms)     # Temp range from 15°C to 42°C
humidity = np.random.uniform(30.0, 95.0, num_farms)        # Humidity percentage from 30% to 95%
soil_moisture = np.random.uniform(10.0, 60.0, num_farms)   # Soil moisture levels from 10% to 60%

# Creating a mathematical relationship to simulate actual crop yield damage percentage (Target Y)
# High temperature and high humidity increase disease spread; good soil moisture keeps plant stable.
raw_damage = (0.6 * temperature) + (0.4 * humidity) - (0.5 * soil_moisture) + np.random.normal(0, 3, num_farms)
crop_damage_percent = np.clip(raw_damage, 0.0, 100.0)      # Force boundaries between 0% and 100%

# 2. Organizing separate NumPy arrays into a clean structured Pandas DataFrame (Table)
agricultural_data = pd.DataFrame({
    'Air_Temperature': temperature,
    'Air_Humidity': humidity,
    'Soil_Moisture_Level': soil_moisture,
    'Actual_Damage_Yield': crop_damage_percent
})

print("\n=== SUCCESS: PANDAS DATAFRAME GENERATED ===")
print("Displaying the top 5 records of your generated dataset:")
print("-" * 65)
print(agricultural_data.head()) # Shows the structured table grid to check NumPy/Pandas integration

# 3. Separating data into inputs matrix (X) and target vector (y)
X = agricultural_data[['Air_Temperature', 'Air_Humidity', 'Soil_Moisture_Level']]
y = agricultural_data['Actual_Damage_Yield']

# 4. Initialize and train the Supervised Linear Regression Model (Executes OLS/Gradient Descent Optimization)
severity_model = LinearRegression()
severity_model.fit(X, y)

print("\n=== SYSTEM LOG ===")
print("Model Status: Linear Regression training completed successfully!")

# 5. Evaluate the model performance manually using .score() and Mean Squared Error (MSE)
y_pred = severity_model.predict(X)
r2_accuracy = severity_model.score(X, y) # Calculates the R² Score (Coefficient of Determination)
mse_value = mean_squared_error(y, y_pred) # Calculates Mean Squared Error

print("-" * 65)
print(f"1. R-Squared (R²) Accuracy Score: {r2_accuracy * 100:.2f}%")
print(f"2. Mean Squared Error (MSE): {mse_value:.4f}")

# 6. Testing the model on a brand new unseen high-risk farm scenario
unseen_farm_metrics = np.array([[38.0, 85.0, 15.0]]) # Temp = 38°C, Humidity = 85%, Moisture = 15%
final_prediction = severity_model.predict(unseen_farm_metrics)

print(f"3. CRITICAL LIVE PREDICTION: Estimated crop destruction score for this farm: {final_prediction[0]:.2f}%")
```
